```python
from pathlib import Path
import csv

from evaluate_zero_shot_detection import EvaluationConfig, run_evaluation

ROOT = Path(r"C:\Users\USER\ZeroShotCleaningAssessment")
DATA_DIR = ROOT / "data"
OUTPUTS = ROOT / "outputs"

ANNOTATIONS = DATA_DIR / "annotations.json"
IMAGE_ROOT = DATA_DIR

LIMIT = None          # для теста: 50 или 200; для финала: None
FORCE_RERUN = False   # True = пересчитать, False = взять metrics.json из outputs
SAVE_PREDICTIONS = True
VERBOSE = True
```

```python
def evaluate_model(title, backend, model, output_name):
    output_dir = OUTPUTS / output_name

    config = EvaluationConfig(
        annotations=ANNOTATIONS,
        image_root=IMAGE_ROOT,
        output_dir=output_dir,
        backend=backend,
        model=model,
        limit=LIMIT,
        save_predictions=SAVE_PREDICTIONS,
        use_cache=not FORCE_RERUN,
        verbose=VERBOSE,
    )

    metrics = run_evaluation(config)
    metrics["title"] = title
    return metrics
```

```python
grounding_dino = evaluate_model(
    title="Grounding DINO Base",
    backend="grounding-dino",
    model="IDEA-Research/grounding-dino-base",
    output_name="zero_shot_detection_grounding_dino_base",
)

grounding_dino
```

```python
yolov8x_world = evaluate_model(
    title="YOLOv8x-World",
    backend="yolo-world",
    model="yolov8x-world.pt",
    output_name="zero_shot_detection_yolov8x_world",
)

yolov8x_world
```

```python
owlvit = evaluate_model(
    title="OWL-ViT Base Patch32",
    backend="owl-vit",
    model="google/owlvit-base-patch32",
    output_name="zero_shot_detection_owlvit_base_patch32",
)

owlvit
```

```python
results = [
    ("YOLOv8x-World", yolov8x_world),
    ("OWL-ViT Base Patch32", owlvit),
    ("Grounding DINO Base", grounding_dino),
]

rows = []

for model_name, metrics in results:
    rows.append({
        "Название модели": model_name,
        "Время обработки датасета (изобр./сек.)": metrics["images_per_second"],
        "Количество обнаруженных боксов": metrics["predicted_boxes"],
        "Precision": metrics["precision"],
        "Recall": metrics["recall"],
        "F1-score": metrics["f1"],
    })

rows
```

```python
def fmt(x):
    return f"{x:.4f}" if isinstance(x, float) else str(x)

headers = list(rows[0])

print("| " + " | ".join(headers) + " |")
print("| " + " | ".join(["---"] * len(headers)) + " |")

for row in rows:
    print("| " + " | ".join(fmt(row[h]) for h in headers) + " |")
```

```python
csv_path = OUTPUTS / "zero_shot_detection_results_table.csv"

with csv_path.open("w", newline="", encoding="utf-8-sig") as f:
    writer = csv.DictWriter(f, fieldnames=list(rows[0]))
    writer.writeheader()
    writer.writerows(rows)

csv_path
```
